1. Imputasi missing value dengan mean, median, dan modus

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Membaca data
data = pd.read_csv('diabetes.csv')

# Menyimpan salinan data
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Membaca data
data = pd.read_csv('diabetes.csv')

# Menampilkan informasi dataset
print(data.info())

# Menampilkan jumlah missing values sebelum imputasi
missing_values = data.isnull().sum()
print("\nMissing values sebelum imputasi:")
print(missing_values[missing_values > 0])

# Menyimpan salinan data asli untuk visualisasi
data_original = data.copy()

# Imputasi dengan mean
data_mean = data.copy()
data_mean.fillna(data_mean.mean(), inplace=True)

# Imputasi dengan median
data_median = data.copy()
data_median.fillna(data_median.median(), inplace=True)

# Imputasi dengan modus (hanya jika ada kolom kategorikal)
data_mode = data.copy()
for column in data_mode.select_dtypes(include=['object']).columns:
    data_mode[column].fillna(data_mode[column].mode()[0], inplace=True)

# Visualisasi distribusi sebelum dan sesudah imputasi
plt.figure(figsize=(15, 8))

# Subplot untuk distribusi asli
plt.subplot(2, 2, 1)
sns.kdeplot(data_original['Glucose'], color='blue', label='Original', fill=True)
plt.title('Distribusi Glucose: Sebelum Imputasi')
plt.xlabel('Glucose Level')
plt.ylabel('Density')
plt.legend()

# Subplot untuk distribusi setelah imputasi mean
plt.subplot(2, 2, 2)
sns.kdeplot(data_mean['Glucose'], color='orange', label='Imputed (Mean)', fill=True)
plt.title('Distribusi Glucose: Imputasi dengan Mean')
plt.xlabel('Glucose Level')
plt.ylabel('Density')
plt.legend()

# Subplot untuk distribusi setelah imputasi median
plt.subplot(2, 2, 3)
sns.kdeplot(data_median['Glucose'], color='green', label='Imputed (Median)', fill=True)
plt.title('Distribusi Glucose: Imputasi dengan Median')
plt.xlabel('Glucose Level')
plt.ylabel('Density')
plt.legend()

# Subplot untuk distribusi setelah imputasi modus
plt.subplot(2, 2, 4)
sns.kdeplot(data_mode['Glucose'], color='purple', label='Imputed (Mode)', fill=True)
plt.title('Distribusi Glucose: Imputasi dengan Modus')
plt.xlabel('Glucose Level')
plt.ylabel('Density')
plt.legend()

plt.tight_layout()
plt.show()

# Cek missing values setelah imputasi
missing_values_after = {
    'Mean Imputation': data_mean.isnull().sum(),
    'Median Imputation': data_median.isnull().sum(),
    'Mode Imputation': data_mode.isnull().sum()
}

print("\nMissing values setelah imputasi:")
for method, values in missing_values_after.items():
    print(f"{method}:")
    print(values[values > 0])

# Membaca data
data = pd.read_csv('diabetes.csv')

# Menampilkan informasi dataset
print(data.info())

# Menampilkan jumlah missing values sebelum imputasi
missing_values = data.isnull().sum()
print("\nMissing values sebelum imputasi:")
print(missing_values[missing_values > 0])

# Menyimpan salinan data asli untuk visualisasi
data_original = data.copy()

# Imputasi dengan mean
data_mean = data.copy()
data_mean.fillna(data_mean.mean(), inplace=True)

# Imputasi dengan median
data_median = data.copy()
data_median.fillna(data_median.median(), inplace=True)

# Imputasi dengan modus (hanya jika ada kolom kategorikal)
data_mode = data.copy()
for column in data_mode.select_dtypes(include=['object']).columns:
    data_mode[column].fillna(data_mode[column].mode()[0], inplace=True)

# Visualisasi distribusi sebelum dan sesudah imputasi
plt.figure(figsize=(15, 8))

# Subplot untuk distribusi asli
plt.subplot(2, 2, 1)
sns.kdeplot(data_original['Glucose'], color='blue', label='Original', fill=True)
plt.title('Distribusi Glucose: Sebelum Imputasi')
plt.xlabel('Glucose Level')
plt.ylabel('Density')
plt.legend()

# Subplot untuk distribusi setelah imputasi mean
plt.subplot(2, 2, 2)
sns.kdeplot(data_mean['Glucose'], color='orange', label='Imputed (Mean)', fill=True)
plt.title('Distribusi Glucose: Imputasi dengan Mean')
plt.xlabel('Glucose Level')
plt.ylabel('Density')
plt.legend()

# Subplot untuk distribusi setelah imputasi median
plt.subplot(2, 2, 3)
sns.kdeplot(data_median['Glucose'], color='green', label='Imputed (Median)', fill=True)
plt.title('Distribusi Glucose: Imputasi dengan Median')
plt.xlabel('Glucose Level')
plt.ylabel('Density')
plt.legend()

# Subplot untuk distribusi setelah imputasi modus
plt.subplot(2, 2, 4)
sns.kdeplot(data_mode['Glucose'], color='purple', label='Imputed (Mode)', fill=True)
plt.title('Distribusi Glucose: Imputasi dengan Modus')
plt.xlabel('Glucose Level')
plt.ylabel('Density')
plt.legend()

plt.tight_layout()
plt.show()
#asli untuk visualisasi
data_original = data.copy()

# Imputasi dengan mean
data_mean = data.copy()
data_mean.fillna(data_mean.mean(), inplace=True)

# Imputasi dengan median
data_median = data.copy()
data_median.fillna(data_median.median(), inplace=True)

# Imputasi dengan modus
data_mode = data.copy()
for column in data_mode.select_dtypes(include=['object']).columns:
    data_mode[column].fillna(data_mode[column].mode()[0], inplace=True)

# Visualisasi distribusi sebelum dan sesudah imputasi
plt.figure(figsize=(15, 8))

# Subplot untuk distribusi asli
plt.subplot(2, 2, 1)
sns.kdeplot(data_original['Glucose'], color='blue', label='Original', fill=True)
plt.title('Distribusi Glucose: Sebelum Imputasi')
plt.xlabel('Glucose Level')
plt.ylabel('Density')
plt.legend()

# Subplot untuk distribusi setelah imputasi mean
plt.subplot(2, 2, 2)
sns.kdeplot(data_mean['Glucose'], color='orange', label='Imputed (Mean)', fill=True)
plt.title('Distribusi Glucose: Imputasi dengan Mean')
plt.xlabel('Glucose Level')
plt.ylabel('Density')
plt.legend()

# Subplot untuk distribusi setelah imputasi median
plt.subplot(2, 2, 3)
sns.kdeplot(data_median['Glucose'], color='green', label='Imputed (Median)', fill=True)
plt.title('Distribusi Glucose: Imputasi dengan Median')
plt.xlabel('Glucose Level')
plt.ylabel('Density')
plt.legend()

# Subplot untuk distribusi setelah imputasi modus
plt.subplot(2, 2, 4)
sns.kdeplot(data_mode['Glucose'], color='purple', label='Imputed (Mode)', fill=True)
plt.title('Distribusi Glucose: Imputasi dengan Modus')
plt.xlabel('Glucose Level')
plt.ylabel('Density')
plt.legend()

plt.tight_layout()
plt.show()
```

2. Cek korelasi antar variabel dengan heatmap

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes

diabetes = load_diabetes()
data = pd.DataFrame(data=diabetes.data, columns=diabetes.feature_names)
data['target'] = diabetes.target

correlation_matrix = data.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm', square=True, cbar_kws={"shrink": .8})
plt.title('Korelasi Antar Variabel pada Dataset Diabetes')
plt.show()
```

**interpretasi :** Kode ini memuat dataset diabetes dari sklearn, kemudian mengonversi data tersebut menjadi DataFrame menggunakan pandas, di mana kolom-kolomnya dinamai sesuai dengan fitur yang ada dan menambahkan kolom target. Selanjutnya, kode menghitung matriks korelasi antar variabel dalam dataset menggunakan metode `.corr()`. Untuk memvisualisasikan hasilnya, sebuah heatmap dibuat menggunakan seaborn, yang menampilkan nilai korelasi dengan anotasi dan palet warna `coolwarm` untuk menandakan hubungan positif dan negatif antara variabel. Ukuran heatmap diatur agar lebih besar, dan judul ditambahkan untuk menjelaskan isi visualisasi. Output dari kode ini adalah grafik heatmap yang menunjukkan hubungan antar variabel dalam dataset diabetes.

3. Lakukan imbalance handling dengan undersampling

```python
import pandas as pd
from imblearn.under_sampling import RandomUnderSampler

data = pd.read_csv('diabetes.csv')
data['Outcome'] = np.where(data['Outcome'] == 1, 'Diabetes', 'No Diabetes')
X, y = data.drop(columns='Outcome'), data['Outcome']
X_resampled, y_resampled = RandomUnderSampler(random_state=42).fit_resample(X, y)


resampled_data = pd.concat([pd.DataFrame(X_resampled, columns=X.columns), pd.Series(y_resampled, name='Outcome')], axis=1)


print("Jumlah data sebelum undersampling:")
print(data['Outcome'].value_counts().to_string())

print("\nJumlah data setelah undersampling:")
print(resampled_data['Outcome'].value_counts().to_string())

print("\nJumlah sampel di DataFrame baru:")
print(resampled_data.shape)
```

**Interpretasi :** Kode ini membaca dataset "diabetes.csv," mengubah kolom Outcome dari nilai numerik (0 dan 1) menjadi kategori "No Diabetes" (untuk 0) dan "Diabetes" (untuk 1). Kemudian, kode memisahkan fitur (kolom selain Outcome) dan target (kolom Outcome) untuk dilakukan proses undersampling menggunakan teknik RandomUnderSampler dari pustaka imbalanced-learn. Teknik ini bertujuan untuk menyeimbangkan jumlah kelas dalam dataset yang awalnya tidak seimbang (jumlah kasus "No Diabetes" lebih banyak dari "Diabetes") dengan mengurangi sampel dari kelas mayoritas sehingga kedua kelas memiliki jumlah yang sama. Setelah itu, data hasil undersampling dikonversi kembali ke dalam bentuk DataFrame. Kode menampilkan jumlah sampel sebelum undersampling, yang menunjukkan ketidakseimbangan antara kelas, dan setelah undersampling, di mana jumlah kelas menjadi seimbang. Terakhir, jumlah total sampel di DataFrame baru juga ditampilkan, menunjukkan ukuran dataset setelah proses undersampling. Dengan output yang akan menampilkan sejumlah data sebelum undersampling (seperti, lebih banyak "Non Diabetes" daripada "Diabates"), kemudian jumlah data setelah undersampling (jumlah "Diabates" dan "Non Diabetes" menjadi sama).

4. Lakukan scaling dengan robust scaler dan minmax

```python
import pandas as pd
from sklearn.preprocessing import RobustScaler, MinMaxScaler


data = pd.read_csv('diabetes.csv')

X = data.drop(columns=['Outcome'])
y = data['Outcome']


robust_scaler = RobustScaler()
minmax_scaler = MinMaxScaler()

X_filled = X.fillna(0)
X_robust_scaled = robust_scaler.fit_transform(X_filled)
X_minmax_scaled = minmax_scaler.fit_transform(X_filled)

df_robust_scaled = pd.DataFrame(X_robust_scaled, columns=X.columns)
df_minmax_scaled = pd.DataFrame(X_minmax_scaled, columns=X.columns)

df_robust_scaled['Outcome'] = y.values
df_minmax_scaled['Outcome'] = y.values

print("Data setelah Robust Scaling:\n", df_robust_scaled.head())
print("\nData setelah MinMax Scaling:\n", df_minmax_scaled.head())
```

**Interpretasi :** Kode ini melakukan scaling pada dataset diabetes menggunakan dua metode: RobustScaler dan MinMaxScaler. Pertama, dataset dibaca dari file CSV dan fitur dipisahkan dari kolom target (Outcome). Kemudian, nilai NaN diisi dengan 0 menggunakan fillna(0) untuk memastikan tidak ada error saat scaling. Selanjutnya, data fitur di-scale dengan RobustScaler, yang mengurangi pengaruh outlier, dan MinMaxScaler, yang menormalkan data ke rentang [0, 1]. Hasil scaling disimpan dalam DataFrame baru untuk masing-masing metode, dan kolom target ditambahkan kembali. Terakhir, dua DataFrame yang berisi data yang sudah di-scale ditampilkan, masing-masing menunjukkan lima baris pertama dari hasil scaling tersebut.
